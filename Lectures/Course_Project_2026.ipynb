## Main purpose

The final project asks you to design and implement a small **causal intelligence pipeline** for a research-relevant problem. The goal is not only to train a model, but to explain how causal assumptions, causal graphs, interventions, counterfactual reasoning, robustness analysis, or causal machine learning can support scientific understanding or decision-making.

A strong project should answer a question of the form:

> What can we learn about a system if we move beyond prediction and reason about causes, interventions, counterfactuals, robustness, or decision consequences?


## Project tracks

Choose **one main track**. You may combine tracks, but the final notebook should have one clearly stated primary focus.

| Track | Focus | Typical question | Example methods |
|---|---|---|---|
| **A** | Causal discovery | What causal structure is plausible from observational or time-series data? | PC, FCI, LiNGAM, NOTEARS, PCMCI / Tigramite |
| **B** | Causal effect estimation | What is the effect of an intervention or treatment on an outcome? | DoWhy, propensity methods, matching, weighting, EconML, DoubleML |
| **C** | Robust and explainable causal AI | Does the model learn mechanisms or shortcuts, and does it remain reliable under shift? | OOD tests, causal explanations, intervention benefit, counterfactual fairness |

**Recommended scope:** Keep the project small, clear, and rigorous. A simple causal analysis with explicit assumptions is better than a complex machine learning pipeline with unclear causal meaning.


## Dataset options

You may use one of the following dataset types:

1. Your own PhD or research dataset.
2. An open-access dataset (e.g. [UC Irvine Machine Learning Repository](https://archive.ics.uci.edu), [OpenTS](https://decisionintelligence.github.io/OpenTS/),...)
3. A synthetic or semi-synthetic dataset generated from a known causal structure.

Your dataset does not need to be large. It does need to be suitable for a causal question. If your real dataset is confidential, you may use an anonymized, aggregated, or synthetic version.

## Deliverable 1 — Dataset and Causal Question Brief

**Due:** After Session 2  
**Length:** Approximately 1 page  

The purpose of this brief is to help you choose a feasible project direction early.

Please include:

- **Project title**  
  A short working title.

- **Dataset**  
  What dataset will you use, and where does it come from?

- **Causal question**  
  What causal question do you want to study?  
  For example: *What is the effect of X on Y?* or *Which variables may causally influence Y?*

- **Main variables**  
  Identify the outcome, the possible treatment/intervention/exposure, and a few important variables.

- **Initial causal idea**  
  Briefly describe your initial DAG, causal graph, temporal structure, or causal assumptions.

- **Project track**  
  Choose one primary track:
  - Track A: Causal discovery
  - Track B: Causal effect estimation
  - Track C: Robust and explainable causal AI

- **Why it is causal**  
  In 2–3 sentences, explain why this is more than a prediction or correlation problem.

**This brief can be preliminary. You may revise your question, variables, and method later.**

## Deliverable 2 — Method Selection and Assumptions Note

**Due:** After Session 4  
**Length:** Approximately 1 page  

The purpose of this note is to help you move from the causal question to a feasible analysis plan.

Please include:

- **Refined causal question**  
  Restate your causal question. It is fine if it has changed since Deliverable 1.

- **Selected method**  
  Name the method or method family you plan to use.  
  Examples: PC, FCI, LiNGAM, NOTEARS, Tigramite, DoWhy, matching, weighting, EconML, DoubleML, OOD testing, or causal explanation methods.

- **Why this method fits**  
  Briefly explain why the method is suitable for your data and question.

- **Main assumptions**  
  List the most important assumptions behind your analysis.  
  Examples: no unmeasured confounding, temporal ordering, causal sufficiency, overlap, stationarity, or invariance.

- **Expected output**  
  What will your analysis produce?  
  For example: a causal graph, an estimated treatment effect, a robustness comparison, or a causal explanation.

- **Possible limitation**  
  Mention one or two limitations you already expect.

**This note is not expected to be perfect. It is a checkpoint to help you prepare the final Colab notebook.**

## Deliverable 3 — Final Submission and Presentation

**Due:** Session 6 — June 10, 2026  
**Format:** Reproducible Google Colab notebook  

In Session 6, each student will present their final Google Colab notebook. You do not need to prepare separate slides.

The final project should be a focused mini-project, not a full research paper. A simple, clear causal analysis is preferred over a complex project with unclear causal meaning.

Your final Colab notebook should include:

- **Project title and track**  
  State your project title and chosen track:
  - Track A: Causal discovery
  - Track B: Causal effect estimation
  - Track C: Robust and explainable causal AI

- **Dataset and motivation**  
  Briefly describe the dataset and why the problem is relevant.

- **Causal question**  
  State the main causal question clearly.

- **Causal assumptions**  
  Show or describe your DAG, causal graph, temporal structure, or main causal assumptions.

- **Method and implementation**  
  Explain which method you use and include the Python code needed to run the analysis.

- **Results**  
  Present the main result using one or more clear figures, tables, or graphs.

- **Causal interpretation**  
  Explain what the result means causally, not only statistically.

- **Limitations**  
  Briefly discuss the main assumptions, uncertainty, or limitations.

- **Conclusion**  
  Summarize the main scientific, engineering, or decision insight.

- **Reproducibility**  
  Make sure the Colab notebook runs from top to bottom.
