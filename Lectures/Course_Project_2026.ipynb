## Main purpose

The final project is a small causal analysis of a research-relevant problem.

The goal is to show how causal thinking can help us understand a system better than prediction alone.

Your project can be guided by a question such as:

> What can we learn about a system if we ask not only what is predicted, but also what may cause what?


## Project tracks

Choose **one main track**. You may combine tracks, but the final notebook should have one clearly stated primary focus.

| Track | Focus | Typical question | Example methods |
|---|---|---|---|
| **A** | Causal discovery | What causal structure is plausible from observational or time-series data? | PC, FCI, LiNGAM, NOTEARS, PCMCI / Tigramite |
| **B** | Causal effect estimation | What is the effect of an intervention or treatment on an outcome? | DoWhy, propensity methods, matching, weighting, EconML, DoubleML |
| **C** | Robust and explainable causal AI | Does the model learn mechanisms or shortcuts, and does it remain reliable under shift? | OOD tests, causal explanations, intervention benefit, counterfactual fairness |

**Recommended scope:** Keep the project small, clear, and rigorous. A simple causal analysis with explicit assumptions is better than a complex machine learning pipeline with unclear causal meaning.


## Dataset options

You may use one of the following dataset types:

1. Your own PhD or research dataset.
2. An open-access dataset (e.g. [UC Irvine Machine Learning Repository](https://archive.ics.uci.edu), [OpenTS](https://decisionintelligence.github.io/OpenTS/),...)
3. A synthetic or semi-synthetic dataset generated from a known causal structure.

Your dataset does not need to be large. It does need to be suitable for a causal question. If your real dataset is confidential, you may use an anonymized, aggregated, or synthetic version.

## Deliverable 1 — Dataset and Methodology Brief

**Due:** Before Session 5 on June 3, 2026 
**Length:** Approximately 1–2 pages  

Please send me a short brief describing your planned final project.

This does not need to be a complete report. It is mainly a checkpoint to help you define a feasible project direction and receive feedback before the final implementation.

In the brief, please describe:

- the dataset you plan to use,
- why the dataset or problem is relevant,
- the main causal question you want to study,
- the main variables you expect to use,
- the project track you plan to follow,
- the method or method family you are considering,
- why this method seems suitable,
- any initial causal assumptions, graph, or temporal reasoning,
- and what you expect the final analysis to produce.

The brief can be preliminary. You may revise your question, method, or dataset after receiving feedback.

The three project tracks are:

- **Track A:** Causal discovery
- **Track B:** Causal effect estimation
- **Track C:** Robust and explainable causal AI

## Deliverable 2 — Final Google Colab Notebook and Presentation

**Due:** Session 6 — final date to be discussed in class  
**Format:** Google Colab notebook  

The final submission is a reproducible Google Colab notebook. In Session 6, you will present your notebook directly from Google Colab. You do not need to prepare separate slides.

The final project should be a focused mini-project, not a full research paper. Please keep the analysis clear, manageable, and well explained.

Your final Colab notebook should include:

- a short project title and selected track,
- a brief description of the dataset and motivation,
- the main causal question,
- the key variables and causal assumptions,
- the method used,
- the Python implementation,
- the main result or visualization,
- a causal interpretation of the result,
- a short discussion of assumptions, uncertainty, or limitations,
- and a brief conclusion.

The notebook should run from top to bottom and should include enough markdown explanation for another reader to understand your workflow.

The presentation will be short, approximately **7–10 minutes**. The main purpose is to explain your question, method, result, and what you learned from the causal analysis.
