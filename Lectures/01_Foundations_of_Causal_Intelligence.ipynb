We keep one simple question in mind:

> **Why did this happen?**

This question is at the heart of causal thinking. Human beings do not only observe the world; we try to understand it. We ask why a patient recovered, why a machine failed, why electricity demand changed, why a model made a wrong prediction, or why an AI system worked in one environment but failed in another.

Prediction is useful, but prediction alone does not answer “why.”

> **Prediction tells us what is likely to happen.  
> Causality helps us understand what would happen if we act, what would have happened otherwise, and why a system behaves as it does.**

Modern data science and artificial intelligence are highly effective at learning patterns from data. Machine learning models can classify images, forecast demand, detect anomalies, summarize text, and generate code. These capabilities are powerful, but they mainly answer questions about **association and prediction**.

Many scientific, engineering, and decision-making problems require a deeper form of reasoning. We often need to know whether changing one part of a system would change another, why an outcome occurred, and what might have happened under a different action. These questions require **causal intelligence**.

Consider the following examples:

- **Health:** Would a treatment improve patient outcomes if assigned to comparable patients?
- **Energy:** Would electricity demand decrease if a demand-response policy were introduced?
- **Manufacturing:** Did preventive maintenance reduce machine failures, or were lower-risk machines simply more likely to be maintained?
- **Climate and remote sensing:** Would vegetation stress decrease if irrigation, land use, or climate-adaptation policies changed?
- **AI safety:** Will a model remain reliable when deployed in a new environment where historical correlations no longer hold?

> These are not only questions about predicting an outcome from observed data. They are questions about **actions**, **mechanisms**, and **alternative realities**.

To understand the difference, it is useful to distinguish between three levels of reasoning: **prediction**, **intervention**, and **counterfactual reasoning**.

---

## 1.1 Predictive reasoning: What is likely to happen?

A predictive model asks:

$$
P(Y \mid X=x)
$$

This means:

> Among cases where we observe $X=x$, what is the expected behavior of $Y$?

For example, in healthcare:

$$
P(\text{Recovery} \mid \text{Treatment}=1)
$$

asks:

> *Among patients who received the treatment, what proportion recovered?*

This is an **observational** question. It describes a pattern in the data. It may be useful for forecasting, risk scoring, or classification. However, it does not necessarily tell us whether the treatment caused recovery.

The reason is simple: patients who received treatment may differ from patients who did not receive treatment. For example, they may have been more severely ill at baseline. Therefore, the observed recovery rate among treated patients is not automatically the causal effect of treatment.

Prediction answers:

> **What do we expect to observe?**

---

## 1.2 Interventional reasoning: What would happen if we act?

A causal intervention model asks:

$$
P(Y \mid do(X=x))
$$

This means:

> *What would happen to $Y$ if we actively intervened and set $X$ to $x$?*

For example:

$$
P(\text{Recovery} \mid do(\text{Treatment}=1))
$$

asks:

> *What would the recovery probability be if we assigned the treatment?*

This is different from simply observing who received the treatment. The expression $do(X=x)$ means that $X$ is actively set to $x$ by an intervention. Examples include assigning a treatment, changing a policy, modifying a machine setting, adjusting a price, or deploying a new AI system.

In other words:

$$
P(Y \mid X=x)
$$

and

$$
P(Y \mid do(X=x))
$$

are not generally the same.

The first expression describes what we observe among units where $X=x$.  
The second expression describes what would happen if we made $X=x$ happen.

For students who want to understand the **do-operator** in more detail, a useful starting point is Pearl, Glymour, and Jewell’s [*Causal Inference in Statistics: A Primer*](http://bayes.cs.ucla.edu/PRIMER/). The do-operator is also introduced in Judea Pearl and Dana Mackenzie’s [*The Book of Why*](http://bayes.cs.ucla.edu/WHY/).

Intervention answers:

> **What would happen if we act?**

---

## 1.3 Counterfactual reasoning: What would have happened otherwise?

A counterfactual model asks:

$$
P(Y_x \mid X=x', Y=y')
$$

This means:

> *For a unit that actually experienced $X=x'$ and outcome $Y=y'$, what would have happened if $X$ had instead been set to $x$?*

For example:

$$
P(\text{Recovery}_{\text{no treatment}}=1 \mid \text{Treatment}=1, \text{Recovery}=1)
$$

asks:

> *For a patient who received treatment and recovered, would the same patient have recovered without treatment?*

This is an **alternative-reality** question. It is more demanding than prediction or intervention because it asks about the same unit under a condition that did not actually occur.

Counterfactual reasoning is important for:

- explaining individual outcomes,
- evaluating responsibility,
- reasoning about fairness,
- supporting personalized decisions,
- understanding what could have happened under a different action.

Counterfactual reasoning answers:

> **What would have happened otherwise?**

---

## 1.4 Three levels of reasoning

The three forms of reasoning can be summarized as follows:

| Type of reasoning | Main question | Typical use |
|---|---|---|
| Prediction / association | What do we observe? | Forecasting, classification, risk scoring |
| Intervention | What would happen if we act? | Experiments, policies, treatment evaluation, decision-making |
| Counterfactual | What would have happened otherwise? | Explanation, responsibility, fairness, personalized reasoning |

<br/>

The corresponding mathematical forms are:

| Type of reasoning | Mathematical form | Meaning |
|---|---|---|
| Prediction / association | $P(Y \mid X=x)$ | What do we observe about $Y$ among cases where $X=x$? |
| Intervention | $P(Y \mid do(X=x))$ | What would happen to $Y$ if we actively set $X$ to $x$? |
| Counterfactual | $P(Y_x \mid X=x', Y=y')$ | For a unit that actually had $X=x'$ and $Y=y'$, what would have happened if $X$ had been set to $x$? |

<br/>

This progression is the starting point of **causal intelligence**. The goal is not to replace prediction, but to extend it. Prediction helps us see patterns. Causality helps us reason about actions, mechanisms, and alternative outcomes.

The purpose of this chapter is to introduce the language and intuition needed to move from **seeing patterns** to **understanding mechanisms** and **supporting decisions**.

In short:

> **Causal intelligence begins with curiosity, but it becomes powerful when “why?” is translated into formal questions about association, intervention, and counterfactuals.**

</br>

# 2. What is Causality?

A practical definition:

> **Causality is concerned with how a system would change if one part of the system were changed by an intervention.**

This is different from saying that two variables are associated. Two variables can be strongly correlated without one causing the other.

For example:

- Ice cream sales and sunburns may be correlated because both increase in hot weather.
- Patients receiving a stronger treatment may have worse outcomes because they were more severely ill at baseline.
- Highly accurate AI models may rely on spurious features that fail under distribution shift.

> Causal inference tries to answer questions about **cause-and-effect relationships** by combining data with assumptions about how the data were generated.


Causal thinking has a long history in philosophy and science. Aristotle argued that scientific understanding requires explanations of why things happen. David Hume later emphasized that we do not directly observe causation; we observe sequences and regularities. For example, one billiard ball hits another, and the second ball moves. 

- We observe the sequence, but the causal connection is inferred.
- Modern causal inference adds a key idea: **intervention**.
- Passive observation can reveal patterns. Intervention allows us to test whether changing one variable changes another.

A useful teaching analogy is that infants are not only passive observers. They interact with their environment: dropping objects, testing reactions, and repeating actions. In scientific language, they are performing small interventions. Causal intelligence formalizes this spirit of learning through action.

## 3. What is NOT Causality?

## 3.1 Causality is not algebra

If a thermometer reading $Y$ is related to body temperature $X$ by:

$$
Y = kX
$$

the equation alone does not tell us whether body temperature causes the mercury column to rise, or whether the mercury column causes body temperature. Algebra describes a relationship, but it does not by itself encode causal direction.

---
## 3.2 Causality is not only statistics

Statistical concepts are typically expressible in terms of the joint distribution of observed variables:

$$
P(X,Y,Z)
$$

Examples include correlation, regression, conditional probability, likelihood, and statistical dependence.

We are often told that **association is not causation!** However, we forget about it. Therefore, we see numerous spurious/funny correlations like examples in the [Spurious Correlations collection](https://tylervigen.com/spurious-correlations).

  <img src="img/CH01/Spurious_Correlations_Muzzarella.png" alt="Mozzarella Example of Spurious Correlations" width="700"/>
</div>


> Causal concepts generally require information beyond the observed joint distribution. Examples include intervention, confounding, mediation, and counterfactual dependence. We will learn more about them in this course.


For students who want a broader and enjoyable perspective on mathematical thinking, uncertainty, and why statistical patterns can mislead us, Jordan Ellenberg’s book [*How Not to Be Wrong: The Power of Mathematical Thinking*](https://en.wikipedia.org/wiki/How_Not_to_Be_Wrong) is a useful optional reading. 

---
### 3.3. Causality is not ordinary machine learning

Machine learning is extremely powerful for prediction. If the training data and future data come from the same environment, a well-trained model can often make accurate predictions. However, accurate prediction does not always mean causal understanding.

A machine learning model learns patterns in data. Some of these patterns may reflect real causal mechanisms, but others may be shortcuts, proxies, or temporary correlations. Many real-world problems have this challenge.

For example:

- A hospital model may predict patient risk partly from the hospital department or admission procedure, not only from the patient’s medical condition.
- An energy-demand model may learn seasonal patterns, but fail when a new demand-response policy changes consumer behavior.
- A manufacturing model may predict defects using machine settings that are correlated with older machines, rather than identifying the actual process causes of defects.


This becomes problematic when we move from **prediction** to **decision-making**. Decision makers usually do not only ask:

- *What is likely to happen?*

They also ask:

- *What should we do?* 
- *What would happen if we changed a policy, treatment, price, process, or model behavior?* 
- *Would the same model still work in a new environment?*

These questions are **causal**. They require reasoning about interventions, mechanisms, and stability, not only correlations.

> **Machine learning can help us predict outcomes, but causality helps us understand which actions are likely to change those outcomes.**


<br/><br/>
In other words, machine learning is trapped in the **Plato Cave**. See [Judea Pearl's book, the Book of Why](http://bayes.cs.ucla.edu/WHY/).

# 4. Statistical Inference vs. Causal Inference

Most scientific data analyses have one of two broad goals:

1. **Association and prediction:** identify variables associated with an outcome and predict future observations.
2. **Causality and decision support:** understand what causes or influences an outcome, especially under interventions.

Statistical inference often asks:

$$
E[Y \mid X=x]
$$

This means: among units where $X=x$, what is the expected value of $Y$?

Causal inference asks:

$$
E[Y \mid do(X=x)]
$$

This means: what would the expected value of $Y$ be if we actively set $X$ to $x$?

These are not generally the same quantity.

---

## 4.1 Traditional statistical inference paradigm

In traditional statistical inference, we usually work with data that have already been observed.

Assume there is a real-world system that generates data. We can call this system the **data-generating process**:

$$
M
$$

This process produces an observed dataset:

$$
D
$$

From this dataset, we try to learn or estimate the probability distribution of the observed variables:

$$
P
$$

Once we have data from this distribution, traditional statistical analysis focuses on quantities that can be estimated from the observed data.

For example, we may estimate:

- the average value of a variable,
- the correlation between two variables,
- a regression coefficient,
- the probability of an outcome,
- a conditional expectation such as:

$$
E[Y \mid X=x]
$$

More generally, we can write a statistical target as:

$$
Q(P)
$$

This means that the quantity of interest is a function of the observed probability distribution $P$.

<div style="text-align: center;">
  <img src="img/CH01/Data-Generation-Statistical.png" alt="Traditional Statistical Inference Paradigm" width="500"/>
</div>

### Example: Income and Education

Suppose we observe data about education and income. A statistical question could be:

> *What is the expected income among people with 16 years of education?*

Mathematically, this is:

$$
E[\text{Income} \mid \text{Education}=16]
$$

This is an associational question. It can be estimated from observed data, but it does not by itself tell us what would happen if we increased someone’s education.

---

## 4.2 Causal inference paradigm

Causal inference asks a different type of question. Instead of only asking what we observe in the current system, we ask what would happen if we changed the system.

The original data-generating process is:

$$
M
$$

When we intervene, we modify part of this system. For example, we may assign a treatment, change a policy, adjust a price, introduce a training program, or modify a machine setting.

After the intervention, the system is no longer exactly the same. We can represent the modified system as:

$$
M'
$$

The original system generates the observed distribution:

$$
P
$$

The modified system would generate a new distribution:

$$
P'
$$

The causal target is therefore a quantity defined on the interventional distribution:

$$
Q(P')
$$

<div style="text-align: center;">
  <img src="img/CH01/Data-Generation-Causal.png" alt="Causal Inference Paradigm" width="500"/>
</div>


### Example: Income and Education

Suppose we again study education and income. A causal question could be:

> What would the income distribution be if everyone were assigned 16 years of education?

Mathematically, this is:

$$
E[\text{Income} \mid do(\text{Education}=16)]
$$

This is different from:

$$
E[\text{Income} \mid \text{Education}=16]
$$

The first expression is causal because it refers to an intervention. The second expression is associational because it only conditions on what we observed.

---

## 4.3 The central challenge

In most observational studies, we only observe data from the original system:

$$
D \sim P
$$

We usually do not observe data from the modified system:

$$
D' \sim P'
$$

This creates **the central challenge of causal inference**:

> We want to answer questions about $P'$, but we usually only have data from $P$.

Therefore, causal inference requires more than data alone. It requires a combination of:

- **observed data**, to estimate statistical patterns,
- **causal assumptions**, to connect the observed world to the interventional world,
- **domain knowledge**, to justify which variables cause which,
- **identification methods**, to express causal quantities using observable quantities,
- **estimation methods**, to compute the causal effect from data.

In short:

> **Statistical inference estimates patterns in the world we observed.**  
> **Causal inference estimates what would happen in a world changed by an intervention.**

</br>

# 5. Prediction vs. Intervention

A **prediction question** asks what is likely to happen based on observed patterns in existing data.

> Among units with these characteristics, what outcome should we expect?

An **intervention question** asks what would happen if we actively changed something in the system.

> If we take a specific action, how would the outcome change?

These two questions may look similar, but they are fundamentally different. Prediction is about learning from the world as it was observed. Intervention is about reasoning about a world in which we deliberately change one part of the system.

## Example: medical treatment

Suppose we have data from a hospital and want to study whether a treatment improves recovery.

A prediction question is:

$$
P(\text{Recovery} \mid \text{Treatment}=1)
$$

This asks:

> *Among patients who received the treatment, what proportion recovered?*

This is useful for describing the data. However, it does not necessarily tell us whether the treatment caused recovery.

A causal intervention question is:

$$
P(\text{Recovery} \mid do(\text{Treatment}=1))
$$

This asks:

> *What would the recovery probability be if we assigned the treatment to comparable patients?*

The difference is important. In real hospital data, treatment is rarely assigned randomly. More severe patients may be more likely to receive treatment. Less severe patients may recover without treatment. Therefore, treated and untreated patients may differ before treatment even begins.

In that case:

$$
P(Y \mid T=1) \neq P(Y \mid do(T=1))
$$

---
## 5.1 Why the difference matters

If we confuse prediction with intervention, we may make incorrect decisions.

For example, suppose treated patients have lower recovery rates than untreated patients. A purely predictive interpretation may suggest that the treatment is harmful. However, a causal interpretation may reveal that treated patients were initially more severely ill. After accounting for severity, the treatment may actually improve recovery.

This distinction appears in many domains:

- **Health:** Patients receiving a treatment may be sicker at baseline.
- **Energy:** Households joining an energy-saving program may already be more environmentally motivated.
- **Education:** Students receiving tutoring may have lower prior performance.
- **Manufacturing:** Machines selected for maintenance may already show early signs of failure.
- **Business:** Customers receiving discounts may already be at higher risk of leaving.

In each case, the observed comparison may not represent the causal effect of the action.

---
## 5.2 Three kinds of questions

| Type of question | Example | What it tells us | Typical tools |
|---|---|---|---|
| Prediction | Can we predict patient recovery from clinical variables? | What is likely to happen? | Statistical learning, machine learning |
| Intervention | Would treatment improve recovery if assigned? | What would happen if we act? | Experiments, causal inference, adjustment methods |
| Counterfactual | Would this patient have recovered without treatment? | What would have happened otherwise? | SCMs, counterfactual reasoning |

Machine learning is especially strong for prediction. Causal intelligence is needed when the goal is to choose actions, evaluate interventions, explain outcomes, or reason about alternative scenarios.

# 6. Pearl’s Causal Ladder

Judea Pearl’s **Ladder of Causation** in the [Book of Why](http://bayes.cs.ucla.edu/WHY/) provides a useful way to organize different levels of causal reasoning. The ladder helps us distinguish between three types of questions:

1. What can we learn by observing data?
2. What can we learn by intervening in a system?
3. What can we learn by imagining what would have happened under different conditions?

These three levels are often described as **seeing**, **doing**, and **imagining**.

In this section, we use the **same healthcare dataset** for all three rungs of the ladder. This keeps the example consistent and shows how the question changes as we move up the ladder.

## Example: A new treatment and patients recovery

A hospital wants to evaluate whether a new treatment improves patient recovery.

The variables are:

- $S$: baseline severity before treatment,
- $T$: treatment status, where $T=1$ means the patient received treatment,
- $Y$: recovery outcome, where $Y=1$ means the patient recovered.

In the observational hospital data, sicker patients are more likely to receive treatment. **This creates a realistic challenge: treatment assignment is not random.**

<div style="text-align: center;">
  <img src="img/CH01/New-Treatment-Impact-Patients.png" alt="New Treatment Impact on Patients" width="700"/>
</div>

```python
# One healthcare dataset used throughout the causal ladder section

def simulate_healthcare_data(n=3000, seed=42):
    rng = np.random.default_rng(seed)

    # Baseline severity. Higher values mean more severe illness.
    severity = rng.normal(0, 1, n)

    # Treatment assignment is observational:
    # sicker patients are more likely to receive treatment.
    propensity = 1 / (1 + np.exp(-(-0.25 + 1.35 * severity)))
    treatment = rng.binomial(1, propensity, n)

    # Potential recovery probabilities under no treatment and treatment.
    # Treatment helps, while severity reduces recovery.
    p_recovery_no_treatment = 1 / (1 + np.exp(-(0.30 - 1.20 * severity)))
    p_recovery_treatment = 1 / (1 + np.exp(-(0.30 + 0.85 - 1.20 * severity)))

    # Observed recovery depends on actual treatment received.
    recovery_prob = np.where(treatment == 1, p_recovery_treatment, p_recovery_no_treatment)
    recovery = rng.binomial(1, recovery_prob, n)

    return pd.DataFrame({
        "Severity": severity,
        "Treatment": treatment,
        "Recovery": recovery,
        "Propensity": propensity,
        "p_Y0": p_recovery_no_treatment,
        "p_Y1": p_recovery_treatment
    })

health = simulate_healthcare_data()
health.head()
```

```python
# Quick descriptive summary

summary = health.groupby("Treatment").agg(
    patients=("Recovery", "size"),
    mean_severity=("Severity", "mean"),
    recovery_rate=("Recovery", "mean")
)

display(summary)

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

# Severity by treatment group
axes[0].hist(health.loc[health["Treatment"] == 0, "Severity"], bins=35, alpha=0.70,
             label="Untreated", color=COLORS["navy"])
axes[0].hist(health.loc[health["Treatment"] == 1, "Severity"], bins=35, alpha=0.70,
             label="Treated", color=COLORS["orange"])
axes[0].set_title("Baseline severity by treatment group")
axes[0].set_xlabel("Severity")
axes[0].set_ylabel("Number of patients")
axes[0].legend()
clean_axis(axes[0])

# Recovery rates
rates = health.groupby("Treatment")["Recovery"].mean()
axes[1].bar(["Untreated", "Treated"], rates.values, color=[COLORS["navy"], COLORS["orange"]])
axes[1].set_title("Observed recovery rate by treatment group")
axes[1].set_ylabel("Recovery rate")
axes[1].set_ylim(0, 1)
clean_axis(axes[1])

plt.tight_layout()
plt.show()
```

### Interpretation

The table and plots show that treatment assignment is not random in this observational dataset.

The treated patients have a higher average severity score than the untreated patients:

- Untreated group: mean severity is about **-0.49**
- Treated group: mean severity is about **0.57**

This means that patients who received treatment were, on average, more severely ill before treatment.

The observed recovery rates also differ:

- Untreated group: recovery rate is about **0.68**
- Treated group: recovery rate is about **0.58**

At first glance, this may suggest that treated patients recovered less often. However, this comparison is misleading because the two groups are not comparable at baseline. The treated group contains more severe patients, and severity reduces the chance of recovery.

The key lesson is:

> The observed difference in recovery rates is not necessarily the causal effect of treatment.

Before asking whether the treatment works, we must account for baseline severity or use an experimental design such as randomization.

## 6.1 Rung 1: Association — Seeing

The first rung is about observing relationships between variables.

At this level, we ask:

> *If we observe $T=1$, what can we say about $Y$?*

Mathematically:

$$
P(Y=1 \mid T=1)
$$

This is the level of ordinary statistics and much of machine learning. We observe data, find patterns, estimate correlations, and make predictions.

In our healthcare example, this means:

> *Among patients who received the treatment, what fraction recovered?*

This is useful, but it does not answer whether the treatment caused recovery.

```python
# Rung 1: Association

assoc_treated = health.loc[health["Treatment"] == 1, "Recovery"].mean()
assoc_untreated = health.loc[health["Treatment"] == 0, "Recovery"].mean()
assoc_difference = assoc_treated - assoc_untreated

print(f"P(Recovery=1 | Treatment=1): {assoc_treated:.3f}")
print(f"P(Recovery=1 | Treatment=0): {assoc_untreated:.3f}")
print(f"Associational difference:    {assoc_difference:.3f}")

fig, ax = plt.subplots(figsize=(7, 4.5))
ax.bar(["P(Y=1 | T=0)", "P(Y=1 | T=1)"], 
       [assoc_untreated, assoc_treated],
       color=[COLORS["navy"], COLORS["orange"]])
ax.set_title("Rung 1: Association from observed data")
ax.set_ylabel("Observed recovery rate")
ax.set_ylim(0, 1)
clean_axis(ax)
plt.show()
```

### Interpretation

The observed hospital data show:

$$
P(Y=1 \mid T=1) = 0.58
$$

and

$$
P(Y=1 \mid T=0) = 0.68
$$

The naive associational difference is therefore:

$$
P(Y=1 \mid T=1) - P(Y=1 \mid T=0) = -0.09
$$

At first glance, this suggests that treated patients had a lower recovery rate than untreated patients. A purely predictive or associational interpretation might therefore conclude that the treatment is harmful.

However, this conclusion would be misleading. The bar chart only compares patients as they appear in the hospital records. It does not account for the fact that treatment assignment was not random. In this simulated hospital setting, more severe patients were more likely to receive treatment. Therefore, the treated group starts from a worse baseline condition.

The key lesson is:

> The observed recovery rate among treated patients is not the same as the recovery rate we would obtain if treatment were assigned to comparable patients.

Association answers:

> *What do we observe in the data?*

It does not necessarily answer:

> *What would happen if we assigned treatment?*

This is why the first rung of the causal ladder is useful for description and prediction, but not sufficient for causal decision-making.

---
## 6.2 Rung 2: Intervention — Doing

The second rung is about actions and interventions.

At this level, we ask:

> *What would happen to recovery if we actively assigned treatment?*

Mathematically:

$$
P(Y=1 \mid do(T=1))
$$

The do-operator means that we set $T$ to a value by intervention, rather than merely observing that $T$ has that value.

Because this is simulated data, we know the underlying treatment response probabilities. In real observational studies, these probabilities are not known and must be identified using assumptions and estimated from data.

```python
# Rung 2: Intervention

# In this simulated example, we know the potential recovery probabilities.
intervention_do0 = health["p_Y0"].mean()
intervention_do1 = health["p_Y1"].mean()
true_ate = intervention_do1 - intervention_do0

print(f"P(Recovery=1 | do(Treatment=0)): {intervention_do0:.3f}")
print(f"P(Recovery=1 | do(Treatment=1)): {intervention_do1:.3f}")
print(f"True average treatment effect:   {true_ate:.3f}")

# A simple adjusted estimate using logistic regression as an illustration.
X = health[["Treatment", "Severity"]]
y = health["Recovery"]

logit = LogisticRegression(max_iter=1000)
logit.fit(X, y)

health_do1 = health.copy()
health_do1["Treatment"] = 1
health_do0 = health.copy()
health_do0["Treatment"] = 0

adj_do1 = logit.predict_proba(health_do1[["Treatment", "Severity"]])[:, 1].mean()
adj_do0 = logit.predict_proba(health_do0[["Treatment", "Severity"]])[:, 1].mean()
adj_ate = adj_do1 - adj_do0

print(f"\nAdjusted estimate P(Y=1 | do(T=1)): {adj_do1:.3f}")
print(f"Adjusted estimate P(Y=1 | do(T=0)): {adj_do0:.3f}")
print(f"Adjusted estimated effect:          {adj_ate:.3f}")

fig, ax = plt.subplots(figsize=(8, 4.5))
labels = ["Observed difference", "Adjusted estimate", "True intervention effect"]
values = [assoc_difference, adj_ate, true_ate]
ax.bar(labels, values, color=[COLORS["gray"], COLORS["green"], COLORS["orange"]])
ax.axhline(0, color="black", linewidth=0.8)
ax.set_title("Rung 2: From association to intervention")
ax.set_ylabel("Difference in recovery probability")
clean_axis(ax)
plt.xticks(rotation=15)
plt.show()
```

### Interpretation

The numerical results and the bar chart show the same message: the naive observational comparison gives the wrong causal impression.

From the simulated data-generating process, the true interventional recovery probabilities are:

$$
P(Y=1 \mid do(T=0)) = 0.563
$$

$$
P(Y=1 \mid do(T=1)) = 0.716
$$

Therefore, the true **Average Treatment Effect (ATE)** is:

$$
0.716 - 0.563 = 0.152
$$

This means that assigning the treatment increases the recovery probability by about **15.2 percentage points**.

The gray bar in the chart shows the naive observed difference:

$$
P(Y=1 \mid T=1) - P(Y=1 \mid T=0) = -0.099
$$

It is negative because treated patients were more severely ill at baseline. Therefore, the treated and untreated groups are not directly comparable.

The green bar shows the adjusted estimate after controlling for baseline severity:

$$
\widehat{P}(Y=1 \mid do(T=1)) = 0.706
$$

$$
\widehat{P}(Y=1 \mid do(T=0)) = 0.569
$$

So the adjusted estimated effect is:

$$
0.706 - 0.569 = 0.137
$$

The orange bar shows the true intervention effect from the simulation:

$$
0.152
$$

The key lesson is:

> The observed association is negative, but the causal effect is positive.  
> Once we account for baseline severity, the treatment appears beneficial rather than harmful.

This is the second rung of the causal ladder: **intervention**. It asks:

> *What would happen if we act?*

This is the level needed for treatment evaluation, policy analysis, A/B testing, and real-world decision-making.

<br/>

Note: In statistics, the **average treatment effect (ATE)** measures the average causal effect of a treatment or intervention on an outcome. More on ATE is available [here](https://www.andrewheiss.com/blog/2024/03/21/demystifying-ate-att-atu/#average-treatment-effects-ate).

---
## 6.3 Rung 3: Counterfactuals — Imagining

The third rung is about **alternative realities for the same unit**.

At this level, we ask:

> For a specific patient, what would have happened if the treatment decision had been different?

A typical counterfactual query is:

$$
P(Y_0=1 \mid T=1, Y=1)
$$

This asks:

> Among patients who received treatment and recovered, what is the probability that they would also have recovered without treatment?

Here:

- $T=1$ means the patient actually received treatment.
- $Y=1$ means the patient actually recovered.
- $Y_0=1$ means the patient would have recovered under no treatment.

This is different from the intervention question in Rung 2.

In Rung 2, we asked about the average effect of assigning treatment in the population:

$$
P(Y=1 \mid do(T=1))
$$

In Rung 3, we ask about an alternative outcome for the **same patients**:

$$
P(Y_0=1 \mid T=1, Y=1)
$$

This is more demanding because, in real life, we cannot observe both outcomes for the same patient. A patient either receives treatment or does not receive treatment. We cannot directly observe what would have happened to that same patient under the other treatment condition.

In this section, we use the simulated healthcare model to illustrate the idea. Because we created the data-generating process ourselves, the dataset contains two potential recovery probabilities:

- `p_Y1`: the probability of recovery if the patient receives treatment.
- `p_Y0`: the probability of recovery if the patient does not receive treatment.

This allows us to simulate a counterfactual question. For treated patients who recovered, we can compare their factual recovery probability under treatment with their counterfactual recovery probability without treatment.

In real observational data, these counterfactual probabilities are not directly available. They must be inferred using causal assumptions, a causal model, and appropriate estimation methods.

```python
# Rung 3: Counterfactual-style analysis in the simulated dataset

# For treated patients who recovered, estimate their probability of recovery under no treatment.
treated_recovered = health[(health["Treatment"] == 1) & (health["Recovery"] == 1)].copy()

cf_recovery_without_treatment = treated_recovered["p_Y0"].mean()
factual_recovery_probability = treated_recovered["p_Y1"].mean()
individual_benefit_probability = (treated_recovered["p_Y1"] - treated_recovered["p_Y0"]).mean()

print(f"Number of treated patients who recovered: {len(treated_recovered)}")
print(f"Mean factual recovery probability under treatment:      {factual_recovery_probability:.3f}")
print(f"Mean counterfactual recovery probability without treatment: {cf_recovery_without_treatment:.3f}")
print(f"Mean individual-level probability gain from treatment:  {individual_benefit_probability:.3f}")

fig, ax = plt.subplots(figsize=(7, 4.5))
ax.bar(
    ["Factual:\nunder treatment", "Counterfactual:\nwithout treatment"],
    [factual_recovery_probability, cf_recovery_without_treatment],
    color=[COLORS["orange"], COLORS["navy"]]
)
ax.set_title("Rung 3: Counterfactual reasoning for treated recovered patients")
ax.set_ylabel("Mean recovery probability")
ax.set_ylim(0, 1)
clean_axis(ax)
plt.show()
```

### Interpretation

The counterfactual analysis focuses on a specific subgroup:

> Patients who actually received treatment and recovered.

In the simulated dataset, there are **760** such patients.

For these patients, the model estimates that their *average factual recovery probability under treatment* was:

$$
P(Y_1=1 \mid T=1, Y=1) \approx 0.671
$$

The counterfactual question asks what would have happened to these same patients if they had not received treatment. The *estimated average counterfactual recovery probability without treatment* is:

$$
P(Y_0=1 \mid T=1, Y=1) \approx 0.496
$$

The estimated individual-level probability gain from treatment for this subgroup is therefore:

$$
0.671 - 0.496 = 0.175
$$

This means that, for treated patients who recovered, the treatment increased their average probability of recovery by about **17.5 percentage points** compared with the counterfactual scenario in which they did not receive treatment.

The bar chart illustrates this comparison. The orange bar represents **the factual world**, where these patients received treatment. The blue bar represents **the counterfactual world**, where the same patients are evaluated as if they had not received treatment.

This is different from the intervention question in Rung 2. The intervention question asks about the average effect of assigning treatment in the population. The counterfactual question asks about an alternative reality for a specific group of patients.

The key lesson is:

> Counterfactual reasoning compares what actually happened with what would have happened otherwise for the same units.

In real-world data, we cannot directly observe both the factual and counterfactual outcomes for the same patient. This is the **fundamental problem of causal inference**. 

> To answer such questions, we need stronger assumptions and a causal model that connects observed outcomes to possible alternative outcomes.

## 6.4 Summary of the three rungs

| Level | Name | Activity | Query type | Healthcare question |
|---|---|---|---|---|
| 1 | Association | Seeing | Observational | What is the recovery rate among treated patients? |
| 2 | Intervention | Doing | Interventional | What would the recovery rate be if we assigned treatment? |
| 3 | Counterfactual | Imagining | Counterfactual | Would this patient have recovered without treatment? |

The corresponding mathematical forms are:

Association:

$$
P(Y \mid X=x)
$$

Intervention:

$$
P(Y \mid do(X=x))
$$

Counterfactual:

$$
P(Y_x \mid X=x', Y=y')
$$

Each level requires more information than the previous one:

- **Association** can often be learned from passive observation.
- **Intervention** requires assumptions about how actions change the system.
- **Counterfactuals** require a model that can reason about individual-level alternative scenarios.

# 7. Observational vs. Experimental Data

Causal questions can be studied using different types of data. The two most common categories are **observational data** and **experimental data**.

The distinction is important because the way data are generated determines what kinds of causal conclusions we can reasonably make.

---
## 7.1 Observational data

Observational data are collected without the researcher controlling who receives the treatment, intervention, or exposure. We simply observe what happened in the real world.

Examples include:

- hospital records,
- insurance claims,
- energy consumption logs,
- satellite image time series,
- customer transaction data,
- manufacturing sensor data,
- administrative education or employment records.

Observational data are valuable because they often reflect real-world behavior at large scale. However, they also create challenges for causal inference.

For example, in hospital data, patients are not randomly assigned to treatment. Doctors may give treatment to patients who are more severely ill. In energy data, households that join an energy-saving program may already be more environmentally motivated. In manufacturing, machines selected for maintenance may already show early warning signs of failure.

This means that treated and untreated groups may differ before the treatment begins.

The main advantages of observational data are:

- they are often easier and cheaper to collect,
- they may cover large populations,
- they may reflect real-world operating conditions,
- they are useful when experiments are unethical, impossible, or too expensive.

The main limitations are:

- **confounding:** other variables may affect both treatment and outcome,
- **selection bias:** the dataset may not represent the target population,
- **reverse causality:** the outcome may influence the treatment or exposure,
- **measurement bias:** variables may be measured imperfectly,
- **missing variables:** important causes may not be recorded.

In observational studies, we often observe:

$$
P(Y \mid T=t)
$$

but we usually want to know:

$$
P(Y \mid do(T=t))
$$


> The challenge is to bridge this gap using causal assumptions, domain knowledge, and appropriate adjustment methods.


---
## 7.2 Experimental data

Experimental data are generated when the researcher or decision maker actively assigns an intervention. In an experiment, we do not only observe the world; we deliberately change something and measure the outcome.

The key idea is that treatment assignment is controlled by design.

In an ideal experiment, treatment is assigned independently of confounders. This makes the treatment and control groups comparable on average.

In that case:

$$
P(Y \mid do(T=t)) = P(Y \mid T=t)
$$

This equality is powerful. It means that the observed difference between treatment and control groups can be interpreted as a causal effect.

Two common experimental designs are:

1. **Randomized Controlled Trials (RCTs)**
2. **A/B tests**

Both are based on the same causal principle: random assignment creates comparable groups.

## 7.3 Randomized Controlled Trials

A **Randomized Controlled Trial (RCT)** is an experiment in which units are randomly assigned to a treatment group or a control group.

The units may be:

- patients,
- students,
- households,
- machines,
- buildings,
- farms,
- customers,
- regions.

The treatment group receives the intervention. The control group does not receive the intervention, or receives a standard alternative.

### Example: healthcare treatment

Suppose a hospital wants to test whether a new treatment improves recovery.

- **Treatment group:** patients assigned to receive the new treatment.
- **Control group:** patients assigned to receive standard care.
- **Outcome:** whether the patient recovers.

Because treatment is randomly assigned, patient severity, age, prior health status, and other baseline characteristics should be balanced across groups on average.

This helps isolate the causal effect of the treatment.

The causal effect can be estimated as:

$$
E[Y \mid T=1] - E[Y \mid T=0]
$$

and, under random assignment, this estimates:

$$
E[Y \mid do(T=1)] - E[Y \mid do(T=0)]
$$

### Why randomization matters

Randomization breaks the link between treatment assignment and confounders.

In observational hospital data, sicker patients may be more likely to receive treatment. In an RCT, treatment assignment is decided randomly, not by severity. Therefore, severity should be similar in both groups on average.

This is why RCTs are often called the **gold standard** for causal inference.

### Limitations of RCTs

RCTs are powerful, but they are not always possible.

They may be limited by:

- **ethics:** we cannot randomly expose people to harmful conditions,
- **cost:** large experiments can be expensive,
- **time:** outcomes may take years to observe,
- **generalizability:** trial participants may not represent the real-world population,
- **compliance:** some participants may not follow the assigned treatment,
- **spillover effects:** treatment assigned to one unit may affect another unit.

For these reasons, causal inference often combines experimental evidence, observational data, and domain knowledge.


---
## 7.4 A/B Testing

An **A/B test** is a randomized experiment commonly used in digital platforms, business, product development, marketing, and user-experience design.

In an A/B test, users or units are randomly assigned to one of two or more versions:

- **Version A:** the current version, often called the control.
- **Version B:** the new version, often called the treatment.

The goal is to test whether the change causes an improvement in a target outcome.

### Example: website design

Suppose a company wants to know whether a new homepage design increases sign-ups.

- **Version A:** current homepage.
- **Version B:** new homepage.
- **Outcome:** whether the user signs up.

The causal question is:

$$
P(\text{Sign-up}=1 \mid do(\text{Version}=B))
-
P(\text{Sign-up}=1 \mid do(\text{Version}=A))
$$

Because users are randomly assigned to version A or B, the difference in sign-up rates can be interpreted as the causal effect of the new design.

### A/B tests as do-operator examples

In causal notation, assigning a user to version B is an intervention:

$$
do(\text{Version}=B)
$$

Assigning a user to version A is another intervention:

$$
do(\text{Version}=A)
$$

The experiment compares outcomes under these two interventions. We are not merely observing which users happened to see which version. We actively assign the version and then measure the outcome.


### Limitations of A/B testing

A/B tests are useful, but they also have limitations:

- **Short-term outcomes may not reflect long-term effects.**
- **Users may interfere with each other**, for example through social sharing.
- **Results may depend on the tested population**, platform, or time period.
- **Many simultaneous tests can create false discoveries.**
- **A statistically significant effect may not be practically important.**
- **Ethical and privacy constraints may limit what can be tested.**

Therefore, A/B tests should be designed carefully and interpreted in context.


---
## 7.5 RCTs and A/B tests in the causal ladder

Both RCTs and A/B tests belong to the second rung of Pearl’s causal ladder: **intervention**.

They help answer questions of the form:

$$
P(Y \mid do(T=t))
$$

They are stronger than purely observational comparisons because treatment is assigned by design.

However, experiments do not automatically answer every causal question. Counterfactual questions about individual units, long-term effects, spillovers, fairness, and mechanisms may require additional modeling and assumptions.

The main lesson is:

> Experimental data are powerful because they create controlled interventions. Observational data are valuable because they reflect the real world. Causal intelligence requires understanding the strengths and limitations of both.

---
### Example: Simulating an RCT for the healthcare treatment problem

We now return to the same healthcare story used earlier in the chapter. A hospital wants to evaluate whether a treatment improves patient recovery.

In the observational dataset, treatment assignment depended on baseline severity: sicker patients were more likely to receive treatment. This made the treated and untreated groups difficult to compare directly.

In this example, we simulate an ideal **Randomized Controlled Trial (RCT)**. The key difference is that treatment is assigned randomly, not based on severity.

The simulation follows four steps:

1. Generate a baseline severity score for each patient.
2. Randomly assign each patient to treatment or control with probability 0.5.
3. Generate recovery probabilities for each patient under no treatment and treatment.
4. Compare recovery rates between the randomized treatment and control groups.

The main idea is that randomization should balance severity across the two groups. If severity is balanced, then the difference in recovery rates can be interpreted as an estimate of the causal treatment effect.

In causal notation, the RCT helps us estimate:

$$
P(Y=1 \mid do(T=1)) - P(Y=1 \mid do(T=0))
$$

using the observed difference:

$$
P(Y=1 \mid T=1) - P(Y=1 \mid T=0)
$$

because treatment assignment is randomized.

```python
# Simulate an ideal randomized controlled trial for the same healthcare story

def simulate_healthcare_rct(n=3000, seed=7):
    rng = np.random.default_rng(seed)

    severity = rng.normal(0, 1, n)

    # Treatment is randomized, not based on severity.
    treatment = rng.binomial(1, 0.5, n)

    p0 = 1 / (1 + np.exp(-(0.30 - 1.20 * severity)))
    p1 = 1 / (1 + np.exp(-(0.30 + 0.85 - 1.20 * severity)))
    recovery_prob = np.where(treatment == 1, p1, p0)
    recovery = rng.binomial(1, recovery_prob, n)

    return pd.DataFrame({
        "Severity": severity,
        "Treatment": treatment,
        "Recovery": recovery
    })

rct = simulate_healthcare_rct()

rct_summary = rct.groupby("Treatment").agg(
    patients=("Recovery", "size"),
    mean_severity=("Severity", "mean"),
    recovery_rate=("Recovery", "mean")
)

display(rct_summary)

rct_effect = (
    rct.loc[rct["Treatment"] == 1, "Recovery"].mean()
    - rct.loc[rct["Treatment"] == 0, "Recovery"].mean()
)

print(f"Estimated effect from randomized data: {rct_effect:.3f}")

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

axes[0].hist(rct.loc[rct["Treatment"] == 0, "Severity"], bins=35, alpha=0.70,
             label="Untreated", color=COLORS["navy"])
axes[0].hist(rct.loc[rct["Treatment"] == 1, "Severity"], bins=35, alpha=0.70,
             label="Treated", color=COLORS["orange"])
axes[0].set_title("RCT: severity is balanced by randomization")
axes[0].set_xlabel("Severity")
axes[0].set_ylabel("Number of patients")
axes[0].legend()
clean_axis(axes[0])

rates = rct.groupby("Treatment")["Recovery"].mean()
axes[1].bar(["Untreated", "Treated"], rates.values, color=[COLORS["navy"], COLORS["orange"]])
axes[1].set_title("RCT: recovery rate by treatment group")
axes[1].set_ylabel("Recovery rate")
axes[1].set_ylim(0, 1)
clean_axis(axes[1])

plt.tight_layout()
plt.show()
```

### Interpretation

This simulated experiment shows why randomization is so important in causal inference.

In the table, the average severity is almost the same in the two groups:

- Control group: mean severity is about \(-0.04\)
- Treatment group: mean severity is about \(-0.02\)

This means that the treated and untreated patients are very similar in terms of baseline severity. The first plot confirms this visually: the two severity distributions almost overlap.

This is different from the observational case earlier, where sicker patients were more likely to receive treatment. In that case, the treatment group and control group were not directly comparable.

Because treatment is randomized here, the main systematic difference between the two groups is the treatment itself. Therefore, we can compare the recovery rates more fairly.

The second plot shows that:

- Untreated patients have a recovery rate of about \(0.56\)
- Treated patients have a recovery rate of about \(0.73\)

The estimated treatment effect is therefore:

$$
0.73 - 0.56 \approx 0.16
$$

This means that, in this randomized experiment, the treatment increases the recovery probability by about **16 percentage points**.

The key lesson is:

> Randomization makes the treatment and control groups comparable.  
> Once the groups are comparable, the difference in outcomes can be interpreted as a causal effect.

In causal terms, randomization helps us estimate:

$$
P(Y=1 \mid do(T=1)) - P(Y=1 \mid do(T=0))
$$

using the observed difference between the randomized treatment and control groups.

# 8. Bias and Paradoxes in Causal Reasoning

Real-world data are rarely neutral. They are shaped by how people make decisions, how systems operate, how measurements are collected, and how datasets are selected. As a result, statistical patterns in data can sometimes point in the wrong causal direction.

This section introduces two important paradoxes that are especially useful for learning causal reasoning:

1. **Simpson’s paradox**  
   A relationship observed in aggregated data changes or reverses after we account for an important subgroup or confounding variable.

2. **Berkson’s paradox**  
   Two variables that are independent in the general population appear associated after we condition on a selection variable or collider.

We focus on these two paradoxes because they illustrate two fundamental lessons in causal inference:

- **Simpson’s paradox** shows why we must think carefully about confounding and subgroup structure.
- **Berkson’s paradox** shows why conditioning on the wrong variable can create misleading associations.

These paradoxes are not rare mathematical curiosities. Similar problems appear in many applied settings, including healthcare, education, insurance, energy systems, manufacturing, remote sensing, finance, and AI model evaluation.

For example:

- In healthcare, a treatment may appear harmful overall because it was given to more severe patients.
- In education, a tutoring program may appear ineffective because it was assigned to students already struggling.
- In insurance, claims data may represent only reported events, not all risk events.
- In AI evaluation, a model may look reliable on a selected test set but fail in the broader population.

The goal here is not to develop the full graphical theory yet. That will come in Chapter 2. The goal is to understand why causal assumptions are necessary before interpreting patterns in data.

For a broader overview of bias types in health and empirical research, students can explore the [Catalogue of Bias](https://catalogofbias.org/).

---
## 8.1 Confounding and Simpson’s Paradox

Confounding occurs when a third variable influences both the treatment and the outcome. This third variable is called a **confounder**.

A simple confounding structure is:

$$
Z \rightarrow T \rightarrow Y
$$

and

$$
Z \rightarrow Y
$$

Here:

- $T$ is the treatment or action.
- $Y$ is the outcome.
- $Z$ is the confounder.

The variable $Z$ affects the treatment decision $T$ and also affects the outcome $Y$. If we compare $T$ and $Y$ without accounting for $Z$, the observed association between $T$ and $Y$ may be misleading.

In simple terms:

> Confounding can make a treatment look better or worse than it really is because the treated and untreated groups were already different before treatment.

**Simpson’s paradox** is a special case where a trend observed in aggregated data changes or reverses after we stratify by an important variable, such as a confounder.

This means that the overall comparison may suggest one conclusion, while the subgroup comparisons suggest a different conclusion.

---
### Example: Simpson’s paradox in kidney stone treatment

We now consider a classic healthcare example inspired by kidney stone treatment studies.

Suppose there are two treatment options:

- **Treatment A:** open surgery  
- **Treatment B:** less invasive treatment  

At first, we may compare the overall recovery rates of the two treatments and ask:

> *Which treatment has the higher recovery rate?*

However, patients are not all the same. Kidney stones can be **small** or **large**, and stone size strongly affects recovery. Stone size may also influence which treatment a doctor chooses. For example, more difficult cases may be more likely to receive one treatment than the other.

This creates a causal challenge:

- **Stone size affects treatment choice.**
- **Stone size also affects recovery.**
- Therefore, stone size is a potential **confounder**.

The key question is:

> *Are we comparing the treatments fairly, or are we mixing together patients with different levels of severity?*

This example shows how the overall comparison can suggest one conclusion, while the subgroup comparison suggests the opposite.

```python
# Simpson's paradox healthcare example inspired by kidney stone treatments

kidney = pd.DataFrame({
    "Treatment": ["A: Open surgery", "A: Open surgery", "B: Less invasive", "B: Less invasive"],
    "Stone_Size": ["Small", "Large", "Small", "Large"],
    "Recovered": [81, 192, 234, 55],
    "Total": [87, 263, 270, 80]
})

kidney["Recovery_Rate"] = kidney["Recovered"] / kidney["Total"]
display(kidney)

overall = kidney.groupby("Treatment")[["Recovered", "Total"]].sum()
overall["Recovery_Rate"] = overall["Recovered"] / overall["Total"]
display(overall)

# Create visualizations
fig, axes = plt.subplots(1, 3, figsize=(17, 4.8))

# Plot 1: DAG
G = nx.DiGraph()
G.add_edges_from([
    ("Stone Size", "Treatment"),
    ("Stone Size", "Recovery"),
    ("Treatment", "Recovery")
])

pos = {
    "Stone Size": (0, 1),
    "Treatment": (-1, 0),
    "Recovery": (1, 0)
}

nx.draw(
    G, pos,
    ax=axes[0],
    with_labels=True,
    node_size=2800,
    node_color=COLORS["light_blue"],
    edgecolors=COLORS["navy"],
    arrows=True,
    arrowsize=18,
    font_size=10,
    font_weight="bold",
    width=1.8
)
axes[0].set_title("Causal structure")
axes[0].axis("off")

# Plot 2: Subgroup recovery rates
x_labels = kidney["Treatment"] + "\n" + kidney["Stone_Size"]
axes[1].bar(
    x_labels,
    kidney["Recovery_Rate"],
    color=[COLORS["navy"], COLORS["navy"], COLORS["orange"], COLORS["orange"]]
)
axes[1].set_title("Recovery rates by stone size")
axes[1].set_ylabel("Recovery rate")
axes[1].set_ylim(0, 1)
axes[1].tick_params(axis="x", rotation=25)
clean_axis(axes[1])

# Plot 3: Overall recovery rates
axes[2].bar(
    overall.index,
    overall["Recovery_Rate"],
    color=[COLORS["navy"], COLORS["orange"]]
)
axes[2].set_title("Overall recovery rates")
axes[2].set_ylabel("Recovery rate")
axes[2].set_ylim(0, 1)
axes[2].tick_params(axis="x", rotation=15)
clean_axis(axes[2])

plt.tight_layout()
plt.show()
```


### Interpretation

The DAG shows the main causal concern. **Stone size** affects both the treatment choice and the probability of recovery. This means that stone size is a confounder.

The subgroup chart shows that **Treatment A performs better within both stone-size groups**:

- For **small stones**, Treatment $A$ has a recovery rate of about **93.1%**, while Treatment $B$ has about **86.7%**.
- For **large stones**, Treatment $A$ has a recovery rate of about **73.0%**, while Treatment $B$ has about **68.8%**.

So, when we compare patients with the same stone size, Treatment $A$ looks better in both groups.

However, the overall chart gives the opposite impression:

- Treatment $A$ overall recovery rate: **78.0%**
- Treatment $B$ overall recovery rate: **82.6%**

At the overall level, Treatment $B$ appears better. This is the paradox.

The reversal happens because the two treatments were used on different types of patients. Treatment $A$ was used more often for large stones, which are harder cases. Treatment $B$ was used more often for small stones, which are easier cases.

In other words, the overall comparison mixes together patients with different levels of difficulty. Because the treatment groups are not comparable, the overall recovery rate can be misleading.

The key lesson is:

> When treatment groups differ in important baseline characteristics, the overall comparison may not represent the causal effect of treatment.

This is why causal reasoning requires us to ask not only:

> *Which treatment had better outcomes?*

but also:

> *Were the patients comparable before treatment?*

In this example, **stone size** plays the role of a **confounder**. To make a fair comparison, we should compare Treatment $A$ and Treatment $B$ within the same stone-size group, or use an adjustment method that accounts for stone size.

## 8.2 Selection Bias and Berkson’s Paradox

Selection bias occurs when the data we analyze do not represent the full population we care about. This often happens because units enter the dataset through a specific selection process.

For example:

- hospital data include people who were admitted to hospital, not all people in the population;
- insurance claims data include reported claims, not all risky events;
- social media data include users who post or engage, not all people with opinions;
- manufacturing failure logs include machines that failed or triggered alarms, not all operating machines.

This matters because the selection process itself can change the relationships we observe in the data. A pattern that appears inside the selected dataset may not exist in the full population.

A particularly important case is **Berkson’s paradox**. Berkson’s paradox occurs when two variables that are unrelated in the general population appear to be associated after we condition on a common effect, also called a **collider**.

A simple collider structure is:

$$
A \rightarrow C \leftarrow B
$$

Here, both $A$ and $B$ influence $C$. The variable $C$ is called a collider because two arrows point into it.

If we analyze only units for which $C=1$, we are conditioning on the collider. This can create an artificial association between $A$ and $B$, even when there is no direct relationship between them in the full population.

In simple terms:

> Selection can make two unrelated causes look related, because we only observe cases that passed through the same selection mechanism.

---
### Example: COVID-19 severity, fracture severity, and hospitalization

Suppose we are studying patients in a hospital dataset. A person may be hospitalized because of severe COVID-19, a severe bone fracture, or both.

Let:

- **COVID-19 severity** be one possible reason for hospitalization;
- **fracture severity** be another possible reason for hospitalization;
- **hospitalization** be the selection variable that determines who appears in the hospital dataset.

In the general population, COVID-19 severity and fracture severity may be unrelated. Having a severe fracture does not necessarily imply having severe COVID-19.

However, if we only analyze hospitalized patients, we are conditioning on hospitalization. This can create a misleading negative association. Among hospitalized patients, if someone has very severe COVID-19, they may need less severe fracture symptoms to explain why they were hospitalized. Conversely, if someone has a very severe fracture, they may need less severe COVID-19 symptoms to be included in the hospital dataset.

The key point is:

> The relationship we see inside the selected dataset may not reflect the relationship in the full population.

This is why selection bias is especially important in medical datasets, insurance claims, hiring datasets, social media datasets, and AI benchmark datasets.

```python
# Simulate Berkson's paradox / collider bias

rng = np.random.default_rng(42)
n = 5000

covid_severity = rng.random(n)
fracture_severity = rng.random(n)

# Hospitalization is a selection event caused by either high COVID severity or high fracture severity.
hospitalized = (covid_severity + fracture_severity) > 1.2

berkson_df = pd.DataFrame({
    "COVID_Severity": covid_severity,
    "Fracture_Severity": fracture_severity,
    "Hospitalized": hospitalized
})

corr_population = berkson_df[["COVID_Severity", "Fracture_Severity"]].corr().iloc[0, 1]
corr_hospital = berkson_df.loc[
    berkson_df["Hospitalized"],
    ["COVID_Severity", "Fracture_Severity"]
].corr().iloc[0, 1]

print(f"Correlation in full population: {corr_population:.3f}")
print(f"Correlation among hospitalized patients: {corr_hospital:.3f}")

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# Plot 1: DAG
G = nx.DiGraph()
G.add_edges_from([
    ("COVID Severity", "Hospitalization"),
    ("Fracture Severity", "Hospitalization")
])

pos = {
    "COVID Severity": (-1, 0),
    "Fracture Severity": (1, 0),
    "Hospitalization": (0, -1)
}

nx.draw(
    G, pos,
    ax=axes[0],
    with_labels=True,
    node_size=3000,
    node_color=COLORS["light_blue"],
    edgecolors=COLORS["navy"],
    arrows=True,
    arrowsize=18,
    font_size=10,
    font_weight="bold",
    width=1.8
)
axes[0].set_title("Collider structure")
axes[0].axis("off")

# Plot 2: Scatter plot
axes[1].scatter(
    berkson_df.loc[~berkson_df["Hospitalized"], "COVID_Severity"],
    berkson_df.loc[~berkson_df["Hospitalized"], "Fracture_Severity"],
    alpha=0.12,
    label="Not hospitalized",
    color=COLORS["gray"]
)
axes[1].scatter(
    berkson_df.loc[berkson_df["Hospitalized"], "COVID_Severity"],
    berkson_df.loc[berkson_df["Hospitalized"], "Fracture_Severity"],
    alpha=0.35,
    label="Hospitalized",
    color=COLORS["red"]
)
axes[1].set_xlabel("COVID-19 severity")
axes[1].set_ylabel("Fracture severity")
axes[1].set_title("Berkson's paradox: conditioning on hospitalization")
axes[1].legend()
clean_axis(axes[1])

plt.tight_layout()
plt.show()
```

### Interpretation

The DAG on the left shows the causal structure of this example:

$$
\text{COVID Severity} \rightarrow \text{Hospitalization} \leftarrow \text{Fracture Severity}
$$

Hospitalization is a **collider** because it is influenced by both COVID-19 severity and fracture severity.

The scatter plot on the right shows two groups:

- Pink points: people who are **not hospitalized**
- Red points: people who are **hospitalized**

In the full population, COVID-19 severity and fracture severity are almost unrelated. The correlation is close to zero:

$$
\text{Correlation in full population} = 0.010
$$

This means that, before conditioning on hospitalization, knowing a person’s COVID-19 severity tells us almost nothing about their fracture severity.

However, after we look only at hospitalized patients, the correlation becomes strongly negative:

$$
\text{Correlation among hospitalized patients} = -0.516
$$

This negative relationship is not a real causal effect between COVID-19 severity and fracture severity. It is created by the selection process.


*Why does this happen?*

A person can enter the hospital dataset because of severe COVID-19, a severe fracture, or both. Among hospitalized patients, if someone has very severe COVID-19, they do not need a very severe fracture to explain why they were hospitalized. Similarly, if someone has a very severe fracture, they do not need very severe COVID-19 to be included in the hospital dataset.

This creates an artificial trade-off inside the selected group of hospitalized patients.

The key lesson is:

> Conditioning on a selection variable can create a misleading association between variables that are unrelated in the full population.

This is **Berkson’s paradox**. It is especially important in real-world datasets where we only observe selected cases, such as hospital admissions, insurance claims, job applicants, or social media users.


> **Before interpreting a relationship in a dataset, always ask how units entered the dataset and whether the selection process could distort the observed association.**

# 9. Why Causality Matters for Safe, Explainable, and Adaptive AI

So far, this chapter has focused on three ideas:

1. **Association:** What patterns do we observe in data?
2. **Intervention:** What would happen if we changed something?
3. **Bias:** Why can observed patterns be misleading?

These ideas are not only important for classical causal inference. They are also central to modern AI.

Most machine learning models are trained to find patterns that improve predictive accuracy. This is useful, but it can become risky when the model learns patterns that are not stable causes of the outcome.

A model may perform well during training because it uses:

- a real causal feature,
- a confounder,
- a proxy variable,
- a selection artifact,
- a shortcut,
- a spurious correlation.

The problem is that the model does not automatically know which patterns are causal and which patterns are accidental. If the environment changes, the accidental patterns may disappear or reverse.

This is one reason why causal reasoning matters for:

- **safe AI:** models should not fail unexpectedly when deployed;
- **explainable AI:** explanations should identify meaningful drivers, not only predictive features;
- **adaptive AI:** models should remain useful when policies, populations, or environments change;
- **decision-oriented AI:** models should help us decide what action to take, not only predict what is likely to happen.

The key message is:

> A feature can be predictive without being causal.  
> A model can be accurate in one environment but unreliable in another.

This section uses a simple synthetic example to show why this matters.

## 9.1 Example: Shortcut Learning Under Distribution Shift

In this example, we study a common problem in machine learning:

> A model may learn a shortcut that works in the training data but fails when the environment changes.

In machine learning, a **shortcut** is a feature or pattern that helps a model make accurate predictions in the training data but is not part of the real mechanism that causes the outcome. The shortcut may work because of how the data were collected, where the data came from, or which population was observed. However, when the model is used in a new environment, the shortcut may disappear, weaken, or reverse.

For example, a hospital model may learn that patients from a specific department have higher recovery rates. This pattern may be useful in the training data, but the department itself does not cause recovery. It may only be a proxy for patient severity, treatment protocol, hospital resources, or data collection practice.

Suppose we train a model to predict whether a patient will recover.

The model has access to two features:

1. **Clinical signal**  
   This is a stable feature related to the patient’s health condition. For example, it may represent disease severity, immune response, or another clinical measurement. This feature is connected to the real mechanism that affects recovery.

2. **Hospital shortcut**  
   This is a contextual feature that happens to be correlated with recovery in the training data. For example, it may represent hospital department, admission procedure, device type, data collection protocol, or another local artifact. It helps prediction in the training hospital, but it does not directly cause recovery.


In the training environment, the hospital shortcut is strongly correlated with recovery. A machine learning model may therefore rely heavily on it. However, when the model is deployed in a new hospital or after a change in procedure, the shortcut may no longer have the same meaning. It may even reverse.

The question is:

> *Will the model still work if the shortcut changes?*

We compare two models:

| Model | Features used | Expected behavior |
|---|---|---|
| Shortcut model | Clinical signal + hospital shortcut | Very accurate in the training-like environment, but fragile under shift |
| Stable model | Clinical signal only | Less accurate in the original environment, but more stable after shift |

The purpose of this example is to show that the best predictive model in one environment is not always the most reliable model for deployment.

```python
# Synthetic example: shortcut learning under distribution shift

def generate_hospital_environment(n=2000, shortcut_direction=1.0, seed=42):
    """
    Generate a simple hospital dataset.

    ClinicalSignal:
        A stable feature related to the patient's true recovery mechanism.

    Recovery:
        The outcome. It depends on the clinical signal.

    HospitalShortcut:
        A contextual feature that is correlated with recovery in one environment,
        but the correlation can change or reverse in another environment.
    """
    rng = np.random.default_rng(seed)

    # Stable clinical signal: higher values mean better underlying recovery potential.
    clinical_signal = rng.normal(0, 1, n)

    # Recovery depends on the stable clinical signal.
    recovery = (clinical_signal + rng.normal(0, 0.8, n) > 0).astype(int)

    # Shortcut feature:
    # In one environment, it is positively correlated with recovery.
    # In the shifted environment, the relationship reverses.
    hospital_shortcut = shortcut_direction * (2 * recovery - 1) + rng.normal(0, 1, n)

    return pd.DataFrame({
        "ClinicalSignal": clinical_signal,
        "HospitalShortcut": hospital_shortcut,
        "Recovery": recovery
    })


# Training-like environment: shortcut is positively correlated with recovery.
train_env = generate_hospital_environment(shortcut_direction=1.5, seed=1)

# Test environment with the same shortcut pattern.
test_same = generate_hospital_environment(shortcut_direction=1.5, seed=2)

# Shifted environment: shortcut relationship reverses.
test_shifted = generate_hospital_environment(shortcut_direction=-1.5, seed=3)


# Model 1: uses both the stable signal and the shortcut.
shortcut_model = LogisticRegression()
shortcut_model.fit(train_env[["ClinicalSignal", "HospitalShortcut"]], train_env["Recovery"])

# Model 2: uses only the stable clinical signal.
stable_model = LogisticRegression()
stable_model.fit(train_env[["ClinicalSignal"]], train_env["Recovery"])


def model_accuracy(model, data, features):
    return accuracy_score(data["Recovery"], model.predict(data[features]))


results = pd.DataFrame({
    "Model": [
        "Shortcut model",
        "Shortcut model",
        "Stable model",
        "Stable model"
    ],
    "Environment": [
        "Same environment",
        "Shifted environment",
        "Same environment",
        "Shifted environment"
    ],
    "Accuracy": [
        model_accuracy(shortcut_model, test_same, ["ClinicalSignal", "HospitalShortcut"]),
        model_accuracy(shortcut_model, test_shifted, ["ClinicalSignal", "HospitalShortcut"]),
        model_accuracy(stable_model, test_same, ["ClinicalSignal"]),
        model_accuracy(stable_model, test_shifted, ["ClinicalSignal"])
    ]
})

display(results)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 4.8))

axes[0].scatter(
    train_env["ClinicalSignal"],
    train_env["HospitalShortcut"],
    c=train_env["Recovery"],
    alpha=0.35,
    cmap="coolwarm"
)
axes[0].set_xlabel("Stable clinical signal")
axes[0].set_ylabel("Hospital shortcut")
axes[0].set_title("Training environment")
clean_axis(axes[0])

axes[1].scatter(
    test_shifted["ClinicalSignal"],
    test_shifted["HospitalShortcut"],
    c=test_shifted["Recovery"],
    alpha=0.35,
    cmap="coolwarm"
)
axes[1].set_xlabel("Stable clinical signal")
axes[1].set_ylabel("Hospital shortcut")
axes[1].set_title("Shifted environment")
clean_axis(axes[1])

plt.tight_layout()
plt.show()


fig, ax = plt.subplots(figsize=(8, 4.5))

x = np.arange(2)
width = 0.35

same_acc = [
    results[
        (results["Model"] == "Shortcut model") &
        (results["Environment"] == "Same environment")
    ]["Accuracy"].iloc[0],
    results[
        (results["Model"] == "Stable model") &
        (results["Environment"] == "Same environment")
    ]["Accuracy"].iloc[0]
]

shift_acc = [
    results[
        (results["Model"] == "Shortcut model") &
        (results["Environment"] == "Shifted environment")
    ]["Accuracy"].iloc[0],
    results[
        (results["Model"] == "Stable model") &
        (results["Environment"] == "Shifted environment")
    ]["Accuracy"].iloc[0]
]

ax.bar(
    x - width/2,
    same_acc,
    width,
    label="Same environment",
    color=COLORS["navy"]
)

ax.bar(
    x + width/2,
    shift_acc,
    width,
    label="Shifted environment",
    color=COLORS["orange"]
)

ax.set_xticks(x)
ax.set_xticklabels(["Shortcut model", "Stable model"])
ax.set_ylim(0, 1)
ax.set_ylabel("Accuracy")
ax.set_title("Prediction accuracy under distribution shift")
ax.legend()
clean_axis(ax)

plt.show()
```

### Interpretation

The table and plots compare two models under two environments.

The **shortcut model** uses both the stable clinical signal and the hospital shortcut. In the same environment, it performs very well, with accuracy around **0.96**. This happens because the shortcut is strongly aligned with recovery in the training data.

However, in the shifted environment, the shortcut reverses. The pattern that helped the model before now points in the wrong direction. As a result, the shortcut model performs very poorly, with accuracy around **0.17**.

The **stable model** uses only the clinical signal. Its accuracy is lower in the same environment, around **0.78**, but it remains almost unchanged in the shifted environment, also around **0.78**.

The scatter plots explain why. In the training environment, the hospital shortcut separates the two recovery groups clearly. In the shifted environment, the relationship between the shortcut and recovery changes direction.

The key lesson is:

> A shortcut can make a model look very accurate during training, but unreliable after deployment.

This is why prediction alone is not enough. A highly accurate model may rely on patterns that are not stable mechanisms. Causal thinking helps us ask whether the model is learning something meaningful about the system or only exploiting a fragile shortcut.

To reduce the shortcut problem, we should not rely only on random train-test splits. We should also test models across different environments, hospitals, time periods, sensors, or populations. We can also use domain knowledge and causal graphs to identify which features are likely to be stable mechanisms and which features are only proxies or artifacts.

In practice:

> Avoid shortcut learning by testing under distribution shift and by designing models around stable, meaningful, and causally relevant features.
